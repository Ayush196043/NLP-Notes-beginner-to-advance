```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

# **3. Remove Url**

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

# **4. Remove Punctuation**

```python
import string,time
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,'')
    return text
```

```python
text = 'string. With. Punctuation?'
```

```python
start = time.time()
print(remove_punc(text))
time1 = time.time() - start
print(time1*50000)
```

# **5. Chat Word Treatments**

```python
# Hum GitHub repository me rakhi slang.txt file ko internet se Google Colab ke andar download kar rahe hain, aur us downloaded file ka naam slang.txt rakh rahe hain.
!wget -O slang.txt "https://raw.githubusercontent.com/rishabhverma17/sms_slang_translator/master/slang.txt"
```

```python
# Is code ka kaam download ki hui slang.txt file ko Python me open karke uske andar ka data read karna hai.
with open("slang.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# Ye code tumhare project ka sabse important part hai, kyunki ye slang.txt ke raw data ko ek Python dictionary chat_words me convert karta hai. Isi dictionary ko tumhara chat_conversion() function use karega.
chat_words = {}

with open("slang.txt", "r", encoding="utf-8") as file:
    for line in file:
        line = line.strip()

        if "=" in line:
            short, full = line.split("=", 1)
            chat_words[short.strip().upper()] = full.strip()
```

```python
# Jo sentence user dega, usme agar koi chat short form chat_words dictionary me hai, to usko full form se replace karna.
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

```python
chat_conversion('IMHO he is the best')
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
remove_stopwords('probably my all-time favorite movie, a story of selflessness, sacrifice and dedication to a noble cause, but it\'s not preachy or boring. it just never gets old, despite my having seen it some 15 or more times')
```

# **8. Handling Emoji**

1. Remove emoji
2. Replace emoji

```python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji("Loved the movie. It was 😘😘")
```

```python
remove_emoji("Lmao 😂😂")
```

# **10. Stemming**

```python
import nltk
nltk.download('punkt')
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

# **11. Lemmatization**

```python
import nltk
nltk.download('wordnet')
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
```
